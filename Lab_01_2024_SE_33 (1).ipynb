### Task 4
Use indexing to grab the word `"hello"` from the given nested list.

```python
# Task 4
lst = [1, 2, [3, 4], [5, [100, 200, ['hello']], 23, 11], 1, 7]

# Indexing step-by-step:
# lst[3] -> [5, [100, 200, ['hello']], 23, 11]
# lst[3][1] -> [100, 200, ['hello']]
# lst[3][1][2] -> ['hello']
# lst[3][1][2][0] -> 'hello'

val = lst[3][1][2][0]
print("Task 4:", val)
```

### Task 6: What is the main difference between a tuple and a list?

**Answer:**
The primary differences between a list and a tuple in Python are:

1. **Mutability (Modifiability):**
   - **Lists are mutable:** Once a list is created, you can modify, add, remove, or change its elements.
   - **Tuples are immutable:** Once a tuple is created, its elements cannot be changed, added, or removed.

2. **Syntax:**
   - **Lists** are enclosed in square brackets: `[1, 2, 3]`[cite: 1].
   - **Tuples** are enclosed in parentheses: `(1, 2, 3)`[cite: 1].

3. **Performance and Usage:**
   - Tuples are generally faster and use less memory than lists because they are fixed/immutable.
   - Lists are used for collections of items that may change over time, whereas tuples are used for fixed data structures (like coordinates or database records).

```python
# Task 6 Answer:
"""
Main Difference between List and Tuple:
1. Mutability:
   - List mutable hoti hai (yani iske elements ko change, add ya remove kiya ja sakta hai).
   - Tuple immutable hota hai (ek baar banne ke baad isme koi tabdeeli nahi ki ja sakتی).

2. Syntax:
   - List ko square brackets [ ] se banaya jata hai.
   - Tuple ko parentheses ( ) se banaya jata hai.
"""

# Example: List elements can be changed (Mutable)
my_list = [1, 2, 3]
my_list[0] = 10  # Yeh error nahi dega
print("Modified List:", my_list)

# Note: Tuple me aisi assignment karne par 'TypeError' aa jata hai kyunki wo immutable hai.
```

### Task 7

### Task 9
Write a function that counts the number of times the word `"dog"` occurs in a string. Ignore case.

```python
# Task 9 (Updated & Clean)
def count_dog(st):
    # Yeh tareeqa punctuation aur case dono ko ignore kar ke 'dog' count kar lega
    return st.lower().count('dog')

# Test the function
result = count_dog("This dog runs faster than that Dog, and another dog!")
print("Task 9:", result)
```

### Task 10
Write a function `caught_speeding(speed, is_birthday)`: $\le 60$ "No Ticket", 61–80 "Small Ticket", $\ge 81$ "Big Ticket". On a birthday, your speed can be 5 mph higher.

# Task 10
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed = speed - 5  # Birthday allowance of 5 mph
        
    if speed <= 60:
        return "No Ticket"
    elif speed >= 61 and speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

# Test cases
print("Task 10 (Test 1):", caught_speeding(60, False)) # Expected: No Ticket
print("Task 10 (Test 2):", caught_speeding(65, False)) # Expected: Small Ticket
print("Task 10 (Test 3):", caught_speeding(65, True))  # Expected: No Ticket (65 - 5 = 60)
