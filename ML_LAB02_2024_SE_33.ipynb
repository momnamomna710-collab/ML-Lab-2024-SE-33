## Task 3 · Randomness and reproducibility
- **(a)** Set `np.random.seed(2026)`, then generate one random number between 0 and 1.
- **(b)** Generate 25 numbers from the standard normal distribution.
- **(c)** Set `np.random.seed(42)` and roll 10,000 Ludo dice with `np.random.randint`. Count each face with `np.unique(..., return_counts=True)`.
- **(d)** Estimate P(six) and compare it with 1/6.
- **(e)** Count how many times two sixes came in a row (no loops).

```python
# T3: your code here
# (a) Set random seed to 2026, then generate one random number between 0 and 1
np.random.seed(2026)
rand_num = np.random.rand()

# (b) Generate 25 numbers from the standard normal distribution
std_normal = np.random.randn(25)

# (c) Set random seed to 42, roll 10,000 Ludo dice, and count each face
np.random.seed(42)
dice_rolls = np.random.randint(1, 7, size=10000)
faces, counts = np.unique(dice_rolls, return_counts=True)

# (d) Estimate P(six) and compare it with 1/6
prob_six = np.mean(dice_rolls == 6)

# (e) Count how many times two sixes came in a row (no loops)
two_sixes_in_row = np.sum((dice_rolls[:-1] == 6) & (dice_rolls[1:] == 6))

print("T3a:", rand_num)
print("T3b:", std_normal)
print("T3c - Faces:", faces, "| Counts:", counts)
print("T3d - Estimated P(six):", prob_six, "| Theoretical (1/6):", 1/6)
print("T3e - Two sixes in a row:", two_sixes_in_row)
```

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9: Self-contained code for Advanced Operations
import numpy as np
np.random.seed(100)
scores = np.random.randint(40, 100, size=(6, 5))

# (a) Find students (rows) whose average score is greater than 70
row_averages = np.mean(scores, axis=1)
good_students = np.where(row_averages > 70)[0]

# (b) Clip all scores below 50 up to exactly 50 (passing baseline)
clipped_scores = np.clip(scores, 50, 100)

# (c) Find indices where scores are greater than 90
high_scorers = np.argwhere(scores > 90)

print("T9a - Row Averages:", np.round(row_averages, 2), "| Student Indices > 70:", good_students)
print("T9b - Clipped Scores Matrix:\n", clipped_scores)
print("T9c - Indices of Scores > 90:\n", high_scorers)
```
