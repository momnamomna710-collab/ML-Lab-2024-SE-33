### Task 3: Clean the Data

We systematically clean the dataset by resolving missing values, incorrect string formats, incorrect types, duplicates, and inconsistent categorizations.

#### Task 3 (a) - Handle Missing Values

```python
print("T3a Missing values count before cleaning:\n", df.isnull().sum())

# Count original rows
pre_missing_count = len(df)

# Drop records missing Brand or Model
df = df.dropna(subset=['Brand', 'Model'])

# Fill others with 'Unknown'
df['Registered City'] = df['Registered City'].fillna('Unknown')
df['Fuel'] = df['Fuel'].fillna('Unknown')
df['Transaction Type'] = df['Transaction Type'].fillna('Unknown')

print("\nT3a Rows removed:", pre_missing_count - len(df))
```

#### Task 3 (b) - Formatting and Cleaning Strings

```python
# Fix city name spacing and casing
df['Registered City'] = df['Registered City'].astype(str).str.strip().str.title()

# Strip 'Rs' and commas
df['Price'] = df['Price'].astype(str).str.replace('Rs', '', regex=False).str.replace(',', '', regex=False).str.strip()
print("T3b Sample cleaned Price column values:\n", df['Price'].head(3))
```

#### Task 3 (c) - Type Conversions

```python
df['Price'] = df['Price'].astype('int64')
df['Year'] = df['Year'].astype('Int64')
print("T3c Types updated:\n", df[['Price', 'Year']].dtypes)
```

#### Task 3 (d) - Categorical Inconsistencies

```python
cities_before = df['Registered City'].nunique()
df['Registered City'] = df['Registered City'].replace({'Sukkar': 'Sukkur'})
df['Model'] = df['Model'].replace({'Corrolla Altis': 'Corolla Altis'})
cities_after = df['Registered City'].nunique()
print(f"T3d Cities count before: {cities_before}, after: {cities_after}")
```

#### Task 3 (e) - Drop Duplicates

```python
duplicates_count = df.duplicated().sum()
print("T3e Duplicate ads count:", duplicates_count)
df = df.drop_duplicates()
print("T3e Total rows after dropping duplicates:", len(df))
```

#### Task 3 (f) - Filter Invalid Values

```python
# Clean KMs column
df.loc[(df['KMs Driven'] < 100) | (df['KMs Driven'] > 1000000) | (df['KMs Driven'].isin([12345, 123456])), 'KMs Driven'] = np.nan

# Drop fake price advertisements
df = df[~df['Price'].isin([12345678, 87654321])]

# Drop unreasonably expensive Suzuki/Daihatsu cars
df = df[~((df['Brand'].isin(['Suzuki', 'Daihatsu'])) & (df['Price'] > 5000000))]
```

#### Task 3 (g) - Self Check Dimensions

```python
print("T3g Final dataset shape:", df.shape)
```

### Task 4: Analysis

We feature-engineer new metrics and extract statistical trends from the clean dataset.

#### Task 4 (a) & (b) - Feature Engineering (Price_lakh and Age)

#### Task 4 (c) - Top 5 Most Common Brands

#### Task 4 (d) - Top 5 Cities with Most Ads

#### Task 4 (e) - Median Price (Lakhs) by Fuel Type

```python
print("T4e Median Price by Fuel Type (highest first):\n", df.groupby('Fuel')['Price_lakh'].median().sort_values(ascending=False))
```

#### Task 4 (f) - Corolla GLI Median Price Trend (2010–2018)

```python
gli_subset = df[(df['Model'].str.contains('GLI', case=False, na=False)) & (df['Year'] >= 2010) & (df['Year'] <= 2018)]
gli_trend = gli_subset.groupby('Year')['Price_lakh'].median()
print("T4f Corolla GLI Median Price (2010-2018):\n", gli_trend)
```

#### Task 4 (g) - Top 5 Most Expensive Ads

```python
print("T4g Top 5 most expensive ads:\n")
display(df.sort_values(by='Price_lakh', ascending=False)[['Brand', 'Model', 'Year', 'Price_lakh', 'Registered City']].head(5))
```

```python
print("T4h Cash vs Installment Stats:\n")
print(df.groupby('Transaction Type').agg(
    Count=('Price_lakh', 'count'),
    Median_Price=('Price_lakh', 'median')
))
```

## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# T3: your code here
initial_rows = len(df)

# (a) Missing values: count, drop rows with no Brand/Model, fill others, print removed rows
print("Missing values before cleaning:\n", df.isnull().sum())
df = df.dropna(subset=['Brand', 'Model'])
df['Registered City'] = df['Registered City'].fillna('Unknown')
df['Fuel'] = df['Fuel'].fillna('Unknown')
df['Transaction Type'] = df['Transaction Type'].fillna('Unknown')
removed_rows = initial_rows - len(df)
print("T3a Rows removed due to missing Brand/Model:", removed_rows)

# (b) Formatting: clean city names and Price string
df['Registered City'] = df['Registered City'].str.strip().str.title()
df['Price'] = df['Price'].astype(str).str.replace('Rs', '', regex=False).str.replace(',', '', regex=False).str.strip()

# (c) Data types conversion
df['Price'] = df['Price'].astype('int64')
df['Year'] = df['Year'].astype('Int64')

# (d) Inconsistent categories
cities_before = df['Registered City'].nunique()
df['Registered City'] = df['Registered City'].replace({'Sukkar': 'Sukkur'})
df['Model'] = df['Model'].replace({'Corrolla Altis': 'Corolla Altis'})
cities_after = df['Registered City'].nunique()
print(f"T3d Cities count before: {cities_before}, after: {cities_after}")

# (e) Duplicates
dups_count = df.duplicated().sum()
print("T3e Exact duplicate ads count:", dups_count)
df = df.drop_duplicates()
print("T3e New row count after dropping duplicates:", len(df))

# (f) Invalid values in KMs Driven and Price/Brand filters
df.loc[(df['KMs Driven'] < 100) | (df['KMs Driven'] > 1000000) | (df['KMs Driven'].isin([12345, 123456])), 'KMs Driven'] = np.nan
df = df[~df['Price'].isin([12345678, 87654321])]
df = df[~((df['Brand'].isin(['Suzuki', 'Daihatsu'])) & (df['Price'] > 5000000))]

# (g) Self-check final shape
print("T3g Final Cleaned Shape (Must be 20484, 9):", df.shape)
```
